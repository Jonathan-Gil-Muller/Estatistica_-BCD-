```python
# Definir as variáveis
disciplina = "Química"
qtd_alunos = 35
qtd_aprovados = 31

# Processamento: calcular a taxa de aprovação
# taxa aprovação = (alunos aprovados / total de alunos)*100
taxa = round((qtd_aprovados / qtd_alunos)*100, 1)

# Imprimir o resultado
print("A taxa de aprovação é:",taxa,"%")
print(f"A taxa de aprovação da disciplina de {disciplina} é: {taxa}%")
```

## Interagindo com o Python

```python
# Entrada de dados/informações
nome = input("Qual o seu nome?")
N1 = float(input("Qual a sua nota na prova 1?"))
N2 = float(input("Qual a sua nota na prova 2?"))
N3 = float(input("Qual a sua nota na prova 3?"))

# Processamento: calcular média aritmética
media = (N1 + N2 + N3)/3

# Imprimir o resultado
print("\n")
print(f"A média do aluno(a) {nome} é: {media}")
```

## Listas

```python
# Dados: lista de notas
lista_notas = [ N1 , N2 , N3, 5]

# Processamento: cálculo da média
media_lista = round(sum(lista_notas) / len(lista_notas), 1)

# Imprimir resultado
print(f"A média é: {media_lista}")
```

```python
"""
Problema: usuário informa N1, N2 e média de aprovação
e o algoritmo deve retornar a nota mínima necessária para aprovação
"""
# Coletar Dados
N1 = float(input("Qual a sua nota na prova 1?"))
N2 = float(input("Qual a sua nota na prova 2?"))
aprovacao = float(input("Qual a sua nota mínima para aprovação?"))

# Processamento: calcular qual deve ser a N3 mínima
nota_3 = aprovacao * 3 - N1 - N2

# Imprimir o resultado
print(f"Para aprovação a nota mínima para prova 3 é: {nota_3}")
```
