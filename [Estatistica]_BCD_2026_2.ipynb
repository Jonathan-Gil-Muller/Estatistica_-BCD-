# **Medidas de posição**

```python
# Importar as bibliotecas
import pandas as pd
import numpy as np
from scipy.stats import trim_mean
import statistics as st
```

```python
# Carregando a base de dados
df = pd.read_excel('[Estatistica-BCD]_DadosQuestionario_20262.xlsx', sheet_name="DadosTratadosV2")

df.info()
```

```python
# Selecionar as variáveis que serão utilizadas
idade       = df['IDADE']          # melhor meidda é a mediana = 20 anos
cidade      = df['C_RESIDE']       # melhor medida é a moda = Blumenau
livros      = df['LIVROS']         # melhor medida é a mediana = 2 livros
programacao = df['EX_PROGRAMACAO'] # melhor medida é a moda = básico
```

```python
###############################################
######## MEDIDAS DE POSIÇÃO ###################
###############################################

## Média aritmética simples
media_simples1 = idade.mean()    # com função nativa Python
media_simples2 = np.nanmean(idade)  # com biblioteca Numpy
print(f'Média simples idade: {media_simples1:.1f}')

media_livros = np.mean(livros)
print(f'Média livros: {media_livros:.1f}')


## Média ponderada
dados = [7.5, 8.2, 4.6, 6.0]
pesos = [1, 3, 4, 2]
media_ponderada = np.average(dados, weights=pesos)
print(f'Média ponderada: {media_ponderada:.1f}')


## Média aparada
media_aparada = trim_mean(idade, 0.1)
print(f'Média aparada: {media_aparada:.1f}')


## Mediana
mediana1 = idade.median()       # com função nativa do Python
mediana2 = np.nanmedian(idade)  # com função da biblioteca Numpy
print(f'Mediana idade: {mediana2:.1f}')

mediana_livros = np.nanmedian(livros)
print(f'Mediana livros: {mediana_livros:.1f}')

## Moda
moda1 = cidade.mode()    # com função nativa Python
moda2 = st.mode(cidade)  # com biblioteca Statistics
print(f'Moda cidade: {moda2}')

moda_prog = st.mode(programacao)  # com biblioteca Statistics
print(f'Moda Ex. Prog.: {moda_prog}')
```

Vamos definir funções para auxiliar no fichamento das medidas descritivas de uma variável

```python
def descriva_quanti(dataframe, variavel):
  # dataframe: um dataframe de dados
  # variável: 'str' correspondente ao nome da variável no dataset

  # 1. Filtrar a variável no dataset
  coluna = dataframe[variavel]
  # 2. Calcular medidas de posição
  media = round(np.nanmean(coluna), 2)
  mediana = round(np.nanmedian(coluna), 2)
  # 3. Calcular medidas de dispersão
  min = np.min(coluna)
  max = np.max(coluna)
  amp = max - min
  desvpad = round(np.std(coluna), 2)
  cv  = round((desvpad / media) * 100, 2)
  # 4. Organizar os resultados em um Dataframe
  df = pd.DataFrame({
      'Variável': variavel,
      'Média': media,
      'Mediana': mediana,
      'Min': min,
      'Max': max,
      'Amplitude': amp,
      'Desvio Padrão': desvpad,
      'Coeficiente de Variação': cv
  }, index=[0])
  # 5. retornar o Dataframe com os resultados das estatísticas descritivas
  return df
```

```python
est_moradias = descriva_quanti(df_bcd, 'IDADE')
est_moradias
```

# **Medidas de dispersão**

```python
# Importar bibliotecas
import pandas as pd
import numpy as np
```

```python
# Carregando a base de dados
df = pd.read_excel('[Estatistica-BCD]_DadosQuestionario_20262.xlsx', sheet_name="DadosTratadosV2")

df.info()
```

```python
# Selecionar variáveis

livros_1 = df[df['SEMESTRE'] == '2026/1']['LIVROS'].tolist()
livros_2 = df[df['SEMESTRE'] == '2026/2']['LIVROS'].tolist()

print(livros_1)
print(livros_2)
```

```python
##########################
### Medidas de Posição ###
##########################

media_1 = np.mean(livros_1)
mediana_1 = np.median(livros_1)
print(f'Média 1: {media_1:.1f}')
print(f'Mediana 1: {mediana_1:.1f}')


media_2 = np.mean(livros_2)
mediana_2 = np.median(livros_2)
print(f'\nMédia 2: {media_2:.1f}')
print(f'Mediana 2: {mediana_2:.1f}')
```

```python
############################
### Medidas de Dispersão ###
############################

# Amplitude Total
amp_1 = max(livros_1) - min(livros_1)
amp_2 = max(livros_2) - min(livros_2)
print(f'Amplitude 1: {amp_1}')
print(f'Amplitude 2: {amp_2}')

# Desvio Médio
dm_1 = np.mean(np.abs(livros_1 - np.mean(livros_1)))
dm_2 = np.mean(np.abs(livros_2 - np.mean(livros_2)))
print(f'\nDesvio Médio 1: {dm_1:.1f}')
print(f'Desvio Médio 2: {dm_2:.1f}')

# Variância
var_1 = np.var(livros_1, ddof=1) # Variância Amostral
var_2 = np.var(livros_2, ddof=1)
print(f'\nVariância 1: {var_1:.1f}')
print(f'Variância 2: {var_2:.1f}')

# Desvio Padrão
desvpad_1 = np.std(livros_1, ddof=1) # Desvio Padrão Amostral
desvpad_2 = np.std(livros_2, ddof=1)
print(f'\nDesvio Padrão 1: {desvpad_1:.1f}')
print(f'Desvio Padrão 2: {desvpad_2:.1f}')

# Coeficiente de Variação
cv_1 = (desvpad_1 / media_1)*100
cv_2 = (desvpad_2 / media_2)*100
print(f'\nCoeficiente de Variação 1: {cv_1:.1f}%')
print(f'Coeficiente de Variação 2: {cv_2:.1f}%')
```

```python
###########################################
### Medidas de Dispersão - qualitativas ###
###########################################

### Razão de Variância

def rv(df, variavel):
  # df é um dataframe
  # variavel: é uma string (nome da variável)

  dados = df[variavel].value_counts() # conta a frequência de cada categoria da variável
  dados = np.array(dados)
  n = dados.sum()
  f_moda = dados.max()

  rv = 1 - (f_moda / n)

  return rv

### Indice de variação qualitativa - IQV

def iqv(df, variavel):
  # df é um dataframe
  # variavel: é uma string (nome da variável)

  dados = df[variavel].value_counts() # conta a frequência de cada categoria da variável
  dados = np.array(dados)
  n = dados.sum()
  k = len(dados)
  somatorio = np.sum(dados**2)

  iqv = (k * (n**2 - somatorio)) / (n**2 * (k-1))

  return iqv
```

```python
import statistics

variavel = 'C_RESIDE'

moda = statistics.mode(df[variavel])
print(f'Moda: {moda}')
RV = rv(df, variavel)
print(f'RV = {RV.round(2)}')
```

```python
import statistics

variavel = 'C_RESIDE'

moda = statistics.mode(df[variavel])
print(f'Moda: {moda}')

IQV = iqv(df, variavel)
print(f'IQV = {IQV.round(2)}')
```

# **Assimetria e Curtose**

```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

```python
# Gerar 1000 números inteiros aleatórios entre 17 e 60
np.random.seed(42)   # Garante reprodutibilidade
lista = np.random.randint(low=17, high=60, size=1000)
valores = pd.Series(lista)

# Cálculo da Assimetria
assimetria = valores.skew()

# Cálculo da Curtose
curtose = valores.kurtosis()

print(f"Assimetria: {assimetria:.2f}")
print(f"Curtose: {curtose:.2f}")
```

```python
from scipy.stats import gaussian_kde

# Criando um histograma
plt.hist(valores,
         bins=20,
         density=True,
         color='seagreen',
         edgecolor='black',
         alpha=0.5)

# Calculando e desenhando a curva de Densidade (KDE)
kde = gaussian_kde(valores)
eixo_x = np.linspace(min(valores), max(valores), 200) # Cria pontos do início ao fim do gráfico
plt.plot(eixo_x, kde(eixo_x),
         color='darkgreen',
         linewidth=2,
         label='Densidade (KDE)')

# Adicionando as linhas de Média e Mediana (Exatamente igual ao seu)
plt.axvline(valores.mean(), color='red', linestyle='--', label=f'Média: {valores.mean():.2f}')
plt.axvline(valores.median(), color='blue', linestyle='-', label=f'Mediana: {valores.median():.2f}')

# Personalização
plt.title(f'Análise da Forma da Distribuição\nAssimetria: {assimetria:.2f} | Curtose: {curtose:.2f}',
          fontsize=14, fontweight='bold')
plt.xlabel('Valores', fontsize=12)
plt.ylabel('Densidade', fontsize=12)
plt.legend()

plt.tight_layout()

plt.show()
```

```python
"""
Estudo da Assimetria e Curtose
"""

# Gerando dados aleatórios com o Numpy
np.random.seed(42)
n = 1000
min = 17
max = 60

normal   = np.clip(np.random.normal(43, 8, n), min, max)
ass_pos  = np.clip(np.random.gamma(shape=2, scale=10, size=n) + 17, min, max)
ass_neg  = np.clip(max - np.random.gamma(shape=2, scale=10, size=n), min, max)
curt_pos = np.clip(np.random.laplace(43, 5, n), min, max)
curt_neg = np.random.uniform(min, max, n)

# Criando a estrutura da figura
fig, axes = plt.subplots(3, 2, figsize=(12, 15))
axes = axes.flatten() # Transforma a grade em uma lista: axes[0], axes[1]...

# --- GRÁFICO 1: NORMAL ---
ax = axes[0]
ax.hist(normal, bins=12, density=True, color='lightgray', edgecolor='white')
pd.Series(normal).plot(kind='kde', ax=ax, color='seagreen')
ax.axvline(np.mean(normal), color='red', linestyle='--', label='Média')
ax.axvline(np.median(normal), color='blue', label='Mediana')
ax.set_title(f"Normal\nAssimetria: {pd.Series(normal).skew():.2f}")
ax.legend()

# --- GRÁFICO 2: ASSIMETRIA POSITIVA ---
ax = axes[1]
ax.hist(ass_pos, bins=12, density=True, color='lightgray', edgecolor='white')
pd.Series(ass_pos).plot(kind='kde', ax=ax, color='seagreen')
ax.axvline(np.mean(ass_pos), color='red', linestyle='--')
ax.axvline(np.median(ass_pos), color='blue')
ax.set_title(f"Assimetria Positiva\nAssimetria: {pd.Series(ass_pos).skew():.2f}")

# --- GRÁFICO 3: ASSIMETRIA NEGATIVA ---
ax = axes[2]
ax.hist(ass_neg, bins=12, density=True, color='lightgray', edgecolor='white')
pd.Series(ass_neg).plot(kind='kde', ax=ax, color='seagreen')
ax.axvline(np.mean(ass_neg), color='red', linestyle='--')
ax.axvline(np.median(ass_neg), color='blue')
ax.set_title(f"Assimetria Negativa\nAssimetria: {pd.Series(ass_neg).skew():.2f}")

# --- GRÁFICO 4: CURTOSE POSITIVA ---
ax = axes[3]
ax.hist(curt_pos, bins=12, density=True, color='lightgray', edgecolor='white')
pd.Series(curt_pos).plot(kind='kde', ax=ax, color='seagreen')
ax.set_title(f"Curtose Positiva\nCurtose: {pd.Series(curt_pos).kurt():.2f}")

# --- GRÁFICO 5: CURTOSE NEGATIVA ---
ax = axes[4]
ax.hist(curt_neg, bins=12, density=True, color='lightgray', edgecolor='white')
pd.Series(curt_neg).plot(kind='kde', ax=ax, color='seagreen')
ax.set_title(f"Curtose Negativa\nCurtose: {pd.Series(curt_neg).kurt():.2f}")

# --- GRÁFICO 6: CURTOSE NORMAL (MESOCÚRTICA) ---
ax = axes[5]
ax.hist(normal, bins=12, density=True, color='lightgray', edgecolor='white')
pd.Series(normal).plot(kind='kde', ax=ax, color='seagreen')
curt_normal_val = pd.Series(normal).kurt()
ax.set_title(f"Curtose Normal (Mesocúrtica)\nCurtose: {curt_normal_val:.2f}")

plt.tight_layout()
plt.show()
```

# **Amostragem**

```python
import pandas as pd
import numpy as np
```

```python
# 1. Criando uma População Fictícia (N = 500 alunos)
np.random.seed(42)
populacao = pd.DataFrame({
    'ID': range(1, 501),
    'Nota_Final': np.random.normal(7, 1.5, 500).clip(0, 10),
    'Presenca': np.random.randint(70, 100, 500)
    })

print(populacao)
```

## _Amostragem Aleatórria Simples_

```python
 2. Definindo o tamanho da amostra
tamanho_amostra = 50

# 3. Executando a Amostragem Aleatória Simples
amostra_simples = populacao.sample(n=tamanho_amostra, random_state=42)

# 4. Comparando as médias
print(f"Média da Nota na População (N=500): {populacao['Nota_Final'].mean():.2f}")
print(f"Média da Nota na Amostra (n=50): {amostra_simples['Nota_Final'].mean():.2f}")
```

## _Amostragem sistemática_

```python
# 2. Definição do intervalo do passo (k = N / n)
N = len(populacao)
n = 50      # Tamanho da amostra
k = N // n  # // retorna a parte interira da divisão (sem arredondamento)

# 3. Escolha do ponto de partida aleatório
# Deve ser um número entre 0 e o valor do salto (k-1)
ponto_partida = np.random.randint(0, k)

# 4. Seleção da sequência sistemática
# Criamos uma lista de índices: partida, partida + k, partida + 2k +...
indices_sistematicos = np.arange(ponto_partida, N, step=k)
amostra_sistematica = populacao.iloc[indices_sistematicos]

# 5. Comparando as médias
print(f"Ponto de partida sorteado: {ponto_partida}")
print(f"Intervalo de salto (k): {k}")
print(f"Média da Nota na População: {populacao['Nota_Final'].mean():.2f}")
print(f"Média da Nota na Amostra Sistemática: {amostra_sistematica['Nota_Final'].mean():.2f}")
```

## _Amostragem Estratificada_

```python
# 1. Criando a população com estratos de gênero
np.random.seed(42)
n_total = 500

# Vamos simular uma proporção de 50% Mulheres, 42% Homens e 8% Outros
populacao = pd.DataFrame({
    'ID': range(1, n_total + 1),
    'Genero': np.random.choice(['Feminino', 'Masculino', 'Outros'],
                                size=n_total,
                                p=[0.50, 0.42, 0.08]),
    'Nota_Final': np.random.normal(7, 1.5, n_total).clip(0, 10).round(2)
    })

# 2. Executando a Amostragem Estratificada
# Vamos extrair uma amostra de 10% (n = 50 alunos)
amostra_estratificada = populacao.groupby('Genero', group_keys=False).apply(
    lambda x: x.sample(frac=0.10, random_state=42))

# 3. Comparação de proporcionalidade
print("--- Distribuição na POPULAÇÃO (Real) ---")
print(populacao['Genero'].value_counts(normalize=True).round(2))

print("\n--- Distribuição na AMOSTRA (Estratificada) ---")
print(amostra_estratificada['Genero'].value_counts(normalize=True).round(2))

# 4. Comparação das médias
media_pop = populacao['Nota_Final'].mean()
media_amostra = amostra_estratificada['Nota_Final'].mean()

print("\n" + "="*40)
print(f"Média da Nota (População): {media_pop:.2f}")
print(f"Média da Nota (Amostra):   {media_amostra:.2f}")
print("="*40)
```

# **Cálculo do Tamanho da amostra**

```python
import numpy as np
```

## _População **Infinita**_

### Dados **quanti**tativos

```python
def amostra_quantitativa_infinita(Z, desvio_padrao, E):
    """
    Calcula o tamanho da amostra para dados quantitativos em populações infinitas.

    Parâmetros:
    Z (float): Valor crítico da distribuição normal (ex: 1.96 para 95% de confiança).
    desvio_padrao (float): Desvio padrão estimado ou obtido via piloto (sigma).
    E (float): Margem de erro admissível na mesma unidade de medida dos dados.

    Retorna:
    (float): Tamanho amostral arredondado para cima.
    """
    # Elevando o desvio padrão ao quadrado para obter a variância (sigma^2)
    variancia = desvio_padrao ** 2

    # Aplicando a estrutura da fórmula: (Z^2 * sigma^2) / E^2
    n = ((Z ** 2) * variancia) / (E ** 2)

    # Retorna o inteiro imediatamente superior
    return np.ceil(n)
```

```python
# TESTE DA FUNÇÃO

# Definir as variáveis
fator_z = 1.96
sigma = 4
erro = 0.5

# Execução do cálculo
tamanho_n = amostra_quantitativa_infinita(
    Z=fator_z,
    desvio_padrao=sigma,
    E=erro)

print(f"O tamanho amostral mínimo calculado é: {tamanho_n} indivíduos.")
```

### Dados **quali**tativos

```python
def amostra_qualitativa_infinita(Z, p, E):
    """
    Calcula o tamanho da amostra para dados qualitativos (proporções) em populações infinitas.

    Parâmetros:
    Z (float): Valor crítico da distribuição normal (ex: 1.96 para 95% de confiança).
    p (float): Proporção estimada do evento de interesse na população (ex: 0.50 para 50%).
    E (float): Margem de erro admissível expressa em formato decimal (ex: 0.05 para 5%).

    Retorna:
    float: Tamanho amostral mínimo (n) arredondado para cima.
    """
    # Calculando a proporção complementar (fracasso)
    q = 1 - p

    # Aplicando a estrutura da fórmula: (Z^2 * p * q) / E^2
    n = ((Z ** 2) * p * q) / (E ** 2)

    # np.ceil(n) arredonda para cima; .astype(int) converte o float do numpy para inteiro puro
    return np.ceil(n)
```

```python
# TESTE DA FUNÇÃO

# Definindo os parâmetros:
fator_z = 1.96
proporcao_p = 0.50
erro_toleravel = 0.05

# Execução do cálculo
tamanho_n = amostra_qualitativa_infinita(
    Z=fator_z,
    p=proporcao_p,
    E=erro_toleravel)

print(f"O tamanho amostral mínimo calculado é: {tamanho_n} indivíduos.")
```

## _População **Finita**_

### Dados **quanti**tativos

```python
def amostra_quantitativa_finita(N, Z, E, sigma):
    """
    Calcula o tamanho da amostra para dados quantitativos em populações finitas.

    Parâmetros:
    Z (float): Valor crítico da distribuição normal.
    sigma (float): Desvio padrão estimado ou obtido via piloto (sigma).
    E (float): Margem de erro admissível na mesma unidade de medida dos dados.
    N (int): Tamanho total da população.

    Retorna:
    float: Tamanho amostral mínimo (n) arredondado para cima.
    """
    # Elevando o desvio padrão ao quadrado para obter a variância (sigma^2)
    variancia = sigma ** 2

    # Calculando o numerador: Z^2 * sigma^2 * N
    numerador = (Z ** 2) * variancia * N

    # Calculando o denominador: E^2 * (N - 1) + Z^2 * sigma^2
    denominador = (E ** 2) * (N - 1) + (Z ** 2) * variancia

    # Aplicando a estrutura da fórmula
    n = numerador / denominador

    # np.ceil(n) arredonda para cima; .astype(int) converte o float do numpy para inteiro
    return np.ceil(n)
```

```python
# TESTE DA FUNÇÃO

# Definir as variáveis
N = 13500
fator_z = 1.96
sigma = 4
erro = 0.5

# Execução do cálculo
tamanho_n = amostra_quantitativa_finita(N, fator_z, erro, sigma)

print(f"O tamanho amostral mínimo calculado é: {tamanho_n} indivíduos.")
```

### Dados **quali**tativos

```python
def amostra_qualitativa_finita(Z, p, E, N):
    """
    Calcula o tamanho da amostra para dados qualitativos (proporções) em populações finitas.

    Parâmetros:
    Z (float): Valor crítico da distribuição normal.
    p (float): Proporção estimada do evento de interesse na população.
    E (float): Margem de erro percentual admissível expressa em formato decimal.
    N (int): Tamanho total da população.

    Retorna:
    float: Tamanho amostral mínimo (n) arredondado para cima.
    """
    # Calculando a proporção complementar (fracasso)
    q = 1 - p

    # Calculando o numerador: Z^2 * p * q * N
    numerador = (Z ** 2) * p * q * N

    # Calculando o denominador: E^2 * (N - 1) + Z^2 * p * q
    denominador = (E ** 2) * (N - 1) + (Z ** 2) * p * q

    # Aplicando a estrutura da fórmula
    n = numerador / denominador

    # np.ceil(n) arredonda para cima; .astype(int) converte o float do numpy para inteiro
    return np.ceil(n)
```

```python
# TESTE DA FUNÇÃO

# Definindo as variáveis baseadas no planejamento:
populacao = 250000
fator_z = 1.96
proporcao_p = 0.50
erro_toleravel = 0.05

# Execução do cálculo
tamanho_n = amostra_qualitativa_finita(
    Z=fator_z,
    p=proporcao_p,
    E=erro_toleravel,
    N=populacao)

print(f"O tamanho amostral mínimo calculado é: {tamanho_n} indivíduos.")
```
