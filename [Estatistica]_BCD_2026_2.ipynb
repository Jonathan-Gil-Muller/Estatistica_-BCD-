**Problema 1**: elaborar um algoritmo onde o usuário informa duas notas (N1 e N2) e média de aprovação da disciplina e o algoritmo retornar a nota mínima necessária para aprovação.

image.png

```python
# Coletar Dados
N1 = float(input("Qual a sua nota na prova 1?"))
N2 = float(input("Qual a sua nota na prova 2?"))
media = float(input("Qual a sua nota mínima para aprovação?"))
notas = float(input("Quantas notas tem na disciplina?"))

# Processamento: calcular qual deve ser a N3 mínima
N3 = media * notas - N1 - N2

# Imprimir o resultado
print(f"Para aprovação a nota mínima para prova 3 é: {N3}")
```

**Problema 02**: No mercado do Sr. João, o preço de venda dos produtos comercializados são calculados acrescentado 40% nos respectivos preços de custo (lucro do Sr. João).

Crie um código para calcular o preço de custo de um determinado produto que é vendido no mercado do Sr. João

image.png

```python
# Coleta das informações
venda = 2.70
lucro = 40   # margem de lucro em percentual

# Processamento das informações
custo = (100 * venda) / (100 + lucro)
custo_arredondado = round(custo, 2)

# Impressão do resultado
print(f'O preço de custo do produto é R${custo_arredondado}')
```

## Dicionários e Dataframes

```python
dicionario = {
    'Provas':['Prova 1', 'Prova 2', 'Prova 3'],
    'Aluno1':[5.2, 2.4, 8.0],
    'Aluno2':[8.7, 4.6, 5.9],
    'Aluno3':[9.5, 4.0, 7.5]
    }

print(dicionario)
```

```python
# Importar a biblioteca Pandas
# Pandas é um repositório onde estão disponíveis diversas função para manipulação de tabelas (dataframes)
import pandas as pd

# Converter o dicionário em uma estrutura de DataFrame
df = pd.DataFrame(dicionario)

# Visualizar o DataFrame
#print(df)
df.head()
```

```python
# Inserir uma nova coluna com a média de cada prova

df['Media_Provas'] = round(
    (df['Aluno1'] + df['Aluno2'] + df['Aluno3']) / 3,
    1
)

df.head()
```

```python
# Importar a biblioteca Numpy (operações matemáticas)
import numpy as np

media_aluno1 = round(np.mean(df['Aluno1']),1)
media_aluno2 = round(np.mean(df['Aluno2']),1)
media_aluno3 = round(np.mean(df['Aluno3']),1)

print(f'Média do Aluno 1: {media_aluno1}')
print(f'Média do Aluno 2: {media_aluno2}')
print(f'Média do Aluno 3: {media_aluno3}')
```
